```python
import pandas as pd #library/module
import matplotlib.pyplot as plt #library

# Load the dataset
df = pd.read_excel("Mental_Health_Dataset_Version_2 (1).xlsx")
print(pd)
print(type(pd))

# The 7 mental-health/lifestyle questions that are already numeric (Likert-scale, 1-5)
columns = [
    'Over the past 2 weeks, how often have you felt nervous, anxious, or excessively stressed?',
    'How frequently do racing thoughts or repetitive worry delay your sleep or relaxation?',
    'How often do you experience difficulty concentrating or mental slowness during study sessions?',
    'How often over the past month have you felt a lack of energy, interest, or motivation in daily routines?',
    'How frequently do you experience sudden irritability or mental exhaustion by the end of the day?',
    'Rate your overall physical energy level by evening',
    'How often do you experience digestive discomfort or bloating?'
]

# 1. SUMMARY STATISTICS 
print("SUMMARY STATISTICS")
print(df[columns].describe())

print("FIRST 5 ROWS")
print(df.head())

print("\nLAST 5 ROWS")
print(df.tail())

print("\nDATASET INFO")
df.info()

print("\nSHAPE (rows, columns)")
print(df.shape)



# 2. HISTOGRAMS
df[columns].hist(figsize=(20, 12), bins=5, edgecolor='black')
plt.tight_layout(pad = 3.0)
plt.savefig("histograms.png")
plt.show()

# 3. SCATTERPLOTS
plt.figure(figsize=(6, 5))
plt.scatter(df['Over the past 2 weeks, how often have you felt nervous, anxious, or excessively stressed?'],
            df['Rate your overall physical energy level by evening'])
plt.xlabel("Stress level")
plt.ylabel("Evening physical energy level")
plt.title("Stress vs Evening Energy Level")
plt.savefig("scatter_stress_vs_energy.png")
plt.show()

plt.figure(figsize=(6, 5))
plt.scatter(df['How frequently do racing thoughts or repetitive worry delay your sleep or relaxation?'],
            df['How often do you experience difficulty concentrating or mental slowness during study sessions?'])
plt.xlabel("Racing thoughts delaying sleep")
plt.ylabel("Difficulty concentrating")
plt.title("Racing Thoughts vs Concentration Difficulty")
plt.savefig("scatter_worry_vs_concentration.png")
plt.show()
```
